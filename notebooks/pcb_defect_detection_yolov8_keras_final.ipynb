### Modelo Utilizado
Para realizar a detecção destas anomalias em tempo real e com alta precisão, adotei o modelo **YOLOv8 Small** (`yolo_v8_s_backbone_coco`) implementado através da biblioteca **KerasCV**. ### Classes de Defeitos Mapeadas:
Treinei o modelo para localizar e categorizar de forma autônoma os seguintes 6 tipos de defeitos críticos para a integridade física e funcional das PCBs:
1.  **Missing Hole (Furo Ausente)**: Furos de fixação ou de passagem de componentes que não foram perfurados.
2.  **Mouse Bite (Mordida de Rato)**: Erosão na borda de uma trilha condutora, reduzindo a largura útil do circuito.
3.  **Open Circuit (Trilha Aberta)**: Ruptura total ou parcial na continuidade de uma linha condutora.
4.  **Short (Curto-circuito)**: Conexão metálica indesejada e acidental entre duas ou mais trilhas condutoras adjacentes.
5.  **Spur (Espora / Falta de Cobre)**: Pequena protuberância indesejada ou defeito de corrosão projetando-se de uma trilha condutora.
6.  **Spurious Copper (Cobre Espúrio)**: Fragmentos ou ilhas isoladas de cobre deixadas acidentalmente na placa após o processo de corrosão ácida.

## 2. Pipeline de Carregamento e Transformação de Dados



### Processo de Tratamento de Dados e Rótulos

Para garantir a qualidade do treinamento do meu modelo YOLOv8, implementei uma estratégia de processamento estruturada que atua diretamente nas imagens e nos arquivos de anotações originais. Esse fluxo de engenharia de dados é composto por três etapas principais que desenvolvi:

##### A. Extração de Região de Interesse (ROI) das PCBs
As imagens brutas do dataset podem conter fundos indesejados e variações de enquadramento que dificultam o aprendizado de características puras do circuito. Para mitigar isso, implementei a função `extract_pcb_roi` utilizando algoritmos clássicos de Visão Computacional (OpenCV):
1. **Suavização e Detecção de Bordas**: Aplico um filtro gaussiano seguido pelo detector Canny para isolar os contornos estruturais da placa.
2. **Correção de Perspectiva (Homografia)**: Localizo os quatro cantos do maior contorno retangular encontrado e aplico uma transformação de perspectiva (`cv2.warpPerspective`). Isso retifica e centraliza a PCB na imagem antes de submetê-la à rede.

##### B. Conversão e Decodificação de Rótulos (YOLO -> KerasCV)
Os rótulos originais fornecidos pelo dataset PKU-Market-PCB estão no formato padrão do YOLO: coordenadas normalizadas relativas ao centro da caixa delimitadora, largura e altura `[x_center, y_center, width, height]` com valores reais entre $0$ e $1$.

No entanto, o **KerasCV** exige o formato absoluto `XYXY` em pixels para calcular as caixas delimitadoras internamente. Criei a função `parse_yolo_annotation` para ler os arquivos `.txt`, desnormalizar as coordenadas multiplicando-as pelas dimensões físicas (largura e altura) da imagem retificada e reconstruir os cantos superior esquerdo `[x1, y1]` e inferior direito `[x2, y2]` de cada anotação.

##### C. Construção do Pipeline de Dados Otimizado com `tf.data`
Para lidar com as limitações de memória e processar os dados em alto desempenho, estruturei um pipeline com a API `tf.data`:
* **Gerador Dinâmico**: Utilizo um gerador Python que faz a leitura sob demanda das imagens e rótulos do Google Drive para evitar sobrecarregar a memória RAM.
* **Caching e Prefetching**: Adicionei o método `.cache()` para manter os dados em memória/disco após a primeira leitura das imagens processadas e o `.prefetch()` para paralelizar o carregamento dos próximos lotes enquanto a GPU processa o lote atual, maximizando a eficiência de tempo de execução.

```python
import cv2
import numpy as np
from PIL import Image

def extract_pcb_roi(pil_image: Image.Image) -> tuple[Image.Image, np.ndarray | None]:
    """
    Localiza a PCB na imagem bruta, recorta e corrige a perspectiva.
    Retorna a imagem recortada e a matriz de transformação (para remapeamento).
    """
    img_cv = np.array(pil_image.convert("RGB"))
    gray = cv2.cvtColor(img_cv, cv2.COLOR_RGB2GRAY)

    # 1. Filtro gaussiano para atenuar trilhas e furos
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    # 2. Detecção de bordas
    edges = cv2.Canny(blurred, 50, 150)

    # Dilatação para fechar bordas descontínuas
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))
    dilated = cv2.dilate(edges, kernel, iterations=2)

    # 3. Busca de contornos
    contours, _ = cv2.findContours(dilated, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    if not contours:
        return pil_image, None  # Fallback: retorna imagem original se não achar

    # Ordena pelo tamanho da área e pega o maior
    largest_contour = max(contours, key=cv2.contourArea)
    area = cv2.contourArea(largest_contour)

    # Se a área for insignificante, assume que a imagem já é a própria placa
    if area < (img_cv.shape[0] * img_cv.shape[1] * 0.15):
        return pil_image, None

    # Aproximação poligonal
    peri = cv2.arcLength(largest_contour, True)
    approx = cv2.approxPolyDP(largest_contour, 0.02 * peri, True)

    # Se achou 4 cantos, faz correção de perspectiva (Warp)
    if len(approx) == 4:
        pts = approx.reshape(4, 2)

        # Ordenar os 4 pontos: superior-esq, superior-dir, inferior-dir, inferior-esq
        rect = np.zeros((4, 2), dtype="float32")
        s = pts.sum(axis=1)
        rect[0] = pts[np.argmin(s)]
        rect[2] = pts[np.argmax(s)]

        diff = np.diff(pts, axis=1)
        rect[1] = pts[np.argmin(diff)]
        rect[3] = pts[np.argmax(diff)]

        (tl, tr, br, bl) = rect
        width_a = np.sqrt(((br[0] - bl[0]) ** 2) + ((br[1] - bl[1]) ** 2))
        width_b = np.sqrt(((tr[0] - tl[0]) ** 2) + ((tr[1] - tl[1]) ** 2))
        max_w = max(int(width_a), int(width_b))

        height_a = np.sqrt(((tr[0] - br[0]) ** 2) + ((tr[1] - br[1]) ** 2))
        height_b = np.sqrt(((tl[0] - bl[0]) ** 2) + ((tl[1] - bl[1]) ** 2))
        max_h = max(int(height_a), int(height_b))

        dst = np.array([
            [0, 0],
            [max_w - 1, 0],
            [max_w - 1, max_h - 1],
            [0, max_h - 1]
        ], dtype="float32")

        m_matrix = cv2.getPerspectiveTransform(rect, dst)
        warped = cv2.warpPerspective(img_cv, m_matrix, (max_w, max_h))
        return Image.fromarray(warped), m_matrix

    # Caso não forme 4 vértices perfeitos, faz um recorte retangular simples (Bounding Rect)
    x, y, w, h = cv2.boundingRect(largest_contour)
    cropped = img_cv[y:y+h, x:x+w]
    return Image.fromarray(cropped), None
```

O formato YOLO armazena coordenadas normalizadas `[x_center, y_center, width, height]`. Abaixo, construí a função de descodificação para `xyxy` absoluto em pixels, conforme exigido pelo KerasCV.

```python
import os
import glob
import io
import tensorflow as tf
from PIL import Image

# Garante que o caminho base do projeto seja definido se o kernel tiver reiniciado
if 'DRIVE_PROJECT_PATH' not in globals():
    DRIVE_PROJECT_PATH = "/content/drive/MyDrive/PCB_Defect_Detection"

# Redefinição garantida da função de processamento de anotações YOLO
def parse_yolo_annotation(label_path, img_width, img_height):
    """Converte o ficheiro .txt com anotações YOLO em tensores XYXY absolutos."""
    boxes = []
    classes = []

    if os.path.exists(label_path):
        with open(label_path, "r") as f:
            for line in f.readlines():
                parts = line.strip().split()
                if len(parts) == 5:
                    cls_id = int(parts[0])
                    x_center, y_center, w, h = map(float, parts[1:])

                    # Conversão para XYXY absoluto
                    x1 = (x_center - w / 2.0) * img_width
                    y1 = (y_center - h / 2.0) * img_height
                    x2 = (x_center + w / 2.0) * img_width
                    y2 = (y_center + h / 2.0) * img_height

                    boxes.append([x1, y1, x2, y2])
                    classes.append(cls_id)

    if len(boxes) == 0:
        boxes = tf.zeros((0, 4), dtype=tf.float32)
        classes = tf.zeros((0,), dtype=tf.float32)
    else:
        boxes = tf.convert_to_tensor(boxes, dtype=tf.float32)
        classes = tf.convert_to_tensor(classes, dtype=tf.float32)

    return boxes, classes

# Redefinição corrigida da função de carregamento
def load_dataset_split(image_dir, label_dir):
    """Carrega um conjunto de imagens e anotações gerando um tf.data.Dataset."""
    image_paths = sorted(glob.glob(os.path.join(image_dir, "*.[jJ][pP][gG]")) +
                         glob.glob(os.path.join(image_dir, "*.[pP][nN][gG]")))

    def generator():
        for img_path in image_paths:
            base_name = os.path.splitext(os.path.basename(img_path))[0]
            lbl_path = os.path.join(label_dir, f"{base_name}.txt")

            # Ler o conteúdo binário do arquivo
            with open(img_path, 'rb') as f:
                img_bytes = f.read()

            # Abrir como objeto PIL Image utilizando io.BytesIO
            pil_img = Image.open(io.BytesIO(img_bytes))

            # Executar a extração do ROI no objeto PIL.Image
            img_sel, m = extract_pcb_roi(pil_img)

            # Converter o resultado de volta para tensor do TensorFlow
            img_np = np.array(img_sel.convert("RGB"))
            img = tf.convert_to_tensor(img_np, dtype=tf.uint8)

            h = float(img.shape[0])
            w = float(img.shape[1])

            boxes, classes = parse_yolo_annotation(lbl_path, w, h)
            yield img, {"boxes": boxes, "classes": classes}

    ds = tf.data.Dataset.from_generator(
        generator,
        output_signature=(
            tf.TensorSpec(shape=(None, None, 3), dtype=tf.uint8),
            {
                "boxes": tf.TensorSpec(shape=(None, 4), dtype=tf.float32),
                "classes": tf.TensorSpec(shape=(None,), dtype=tf.float32),
            }
        )
    )
    return ds
```

Cada placa PCB pode conter múltiplos defeitos ou nenhum. Utilize `ragged_batch` para agrupar amostras de cardinalidade variável e a camada nativa `keras_cv.layers.Resizing` para manter a proporção com *padding*.

```python
resizing_layer = keras_cv.layers.Resizing(
    IMG_SIZE[0], IMG_SIZE[1], bounding_box_format="xyxy", pad_to_aspect_ratio=True
)

def format_inputs(images, bounding_boxes):
    inputs = {"images": images, "bounding_boxes": bounding_boxes}
    return resizing_layer(inputs)


# Redefinição garantida da função prepare_pipeline
def prepare_pipeline(dataset, batch_size=BATCH_SIZE, is_training=True):
    # Adiciona cache para evitar reprocessamento de I/O em cada época
    dataset = dataset.cache()

    if is_training:
        # O .repeat() garante que o gerador não se esgote entre as épocas de treino
        dataset = dataset.repeat()
        dataset = dataset.shuffle(128)

    dataset = dataset.ragged_batch(batch_size, drop_remainder=is_training)
    dataset = dataset.map(lambda img, bb: format_inputs(img, bb), num_parallel_calls=tf.data.AUTOTUNE)

    return dataset.prefetch(tf.data.AUTOTUNE)

# Caminho apontando diretamente para o seu Google Drive
DATASET_ROOT = os.path.join(DRIVE_PROJECT_PATH, "pcb-dataset/PKU-Market-PCB(Data enhanced version)")

train_img_dir = os.path.join(DATASET_ROOT, "train/images")
train_lbl_dir = os.path.join(DATASET_ROOT, "train/labels")

val_img_dir = os.path.join(DATASET_ROOT, "valid/images")
val_lbl_dir = os.path.join(DATASET_ROOT, "valid/labels")

test_img_dir = os.path.join(DATASET_ROOT, "test/images")
test_lbl_dir = os.path.join(DATASET_ROOT, "test/labels")

# Recriando os pipelines a partir dos arquivos armazenados no Google Drive
train_ds = prepare_pipeline(load_dataset_split(train_img_dir, train_lbl_dir), is_training=True)
val_ds = prepare_pipeline(load_dataset_split(val_img_dir, val_lbl_dir), is_training=False)
test_ds = prepare_pipeline(load_dataset_split(test_img_dir, test_lbl_dir), is_training=False)

print("Sucesso! Os pipelines agora estão lendo os dados diretamente do Google Drive em:")
print(DATASET_ROOT)
```

### 2.1 Análise Estatística de Distribuição e Necessidade de Data Augmentation

Ao executar a contagem das caixas delimitadoras no meu conjunto de treinamento, observei uma distribuição surpreendentemente uniforme entre as 6 classes de defeitos (com cada classe representando de forma estável entre 16% e 17% do total de 11.812 anotações).

No entanto, a necessidade de aplicar estratégias de aumento de dados condicional (*Data Augmentation*) não decorre unicamente de um desequilíbrio numérico puro (frequência absoluta), mas sim de fatores qualitativos de detecção:

1. **Desbalanceamento de Escala**: Defeitos geométricos como o *spur* (espora) e o *mouse_bite* (mordida de rato) são intrinsecamente menores e sutis em comparação com o *missing_hole* (furo ausente). Eles se localizam estritamente nas bordas das trilhas de cobre, o que dificulta muito a tarefa de regressão de caixa pelo modelo YOLOv8.
2. **Dificuldade de Aprendizado (Sensibilidade)**: Na prática, as métricas preliminares indicam que o modelo apresenta menor taxa de Sensibilidade (*Recall*) justamente nessas classes menores, pois o sinal visual delas é facilmente ignorado no processo de amostragem padrão.

Por isso, decidi introduzir um pipeline de **Data Augmentation condicional e dinâmico** focado especificamente nas classes *mouse_bite* (classe ID 1) e *spur* (classe ID 4) para aumentar artificialmente a variabilidade geométrica e a robustez do meu modelo nestas anomalias específicas.

```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Caminho das anotações de treino
train_labels_pattern = os.path.join(train_lbl_dir, "*.txt")
label_files = glob.glob(train_labels_pattern)

# Dicionário para contar a ocorrência de cada classe
class_counts = {i: 0 for i in CLASS_MAPPING.keys()}

total_objects = 0
for file_path in label_files:
    if os.path.exists(file_path):
        with open(file_path, "r") as f:
            for line in f.readlines():
                parts = line.strip().split()
                if len(parts) > 0:
                    try:
                        cls_id = int(parts[0])
                        if cls_id in class_counts:
                            class_counts[cls_id] += 1
                            total_objects += 1
                    except ValueError:
                        continue

# Criando um DataFrame para exibição e análise
data = []
for cls_id, count in class_counts.items():
    class_name = CLASS_MAPPING[cls_id]
    percentage = (count / total_objects) * 100 if total_objects > 0 else 0
    data.append({
        "Class ID": cls_id,
        "Classe": class_name,
        "Quantidade (Frequência)": count,
        "Proporção (%)": f"{percentage:.2f}%"
    })

df_imbalance = pd.DataFrame(data)
print(f"Total de instâncias/defeitos encontrados no treino: {total_objects}\n")
print(df_imbalance.to_string(index=False))

# Plotando o gráfico de distribuição das classes
plt.figure(figsize=(10, 6))
sns.barplot(
    x="Classe",
    y="Quantidade (Frequência)",
    data=df_imbalance,
    palette="viridis",
    hue="Classe",
    legend=False
)
plt.title("Distribuição e Desbalanceamento de Classes no Conjunto de Treino", fontsize=14)
plt.xlabel("Classes de Defeito", fontsize=12)
plt.ylabel("Quantidade de Instâncias (BBoxes)", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Visualização das Imagens e Anotações (Verificação do meu Dataset)

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

def visualize_batch(dataset_pipeline, num_images=4):
    """Pega um batch do dataset preparado e plota as imagens com seus respectivos Bounding Boxes."""
    batch = next(iter(dataset_pipeline))
    images = batch["images"]
    bounding_boxes = batch["bounding_boxes"]

    # Conversão de RaggedTensors se necessário
    if isinstance(images, tf.RaggedTensor):
        images = images.to_tensor()
    if isinstance(bounding_boxes["boxes"], tf.RaggedTensor):
        boxes = bounding_boxes["boxes"].to_tensor()
        classes = bounding_boxes["classes"].to_tensor()
    else:
        boxes = bounding_boxes["boxes"]
        classes = bounding_boxes["classes"]

    plt.figure(figsize=(15, 10))
    for i in range(min(num_images, BATCH_SIZE)):
        ax = plt.subplot(2, 2, i + 1)
        img = images[i].numpy().astype("uint8")
        plt.imshow(img)

        # Desenha cada box presente na imagem
        img_boxes = boxes[i].numpy()
        img_classes = classes[i].numpy()

        for box, cls_id in zip(img_boxes, img_classes):
            # Desconsidera caixas de padding (-1)
            if cls_id == -1 or np.all(box == 0):
                continue

            x1, y1, x2, y2 = box
            width, height = x2 - x1, y2 - y1

            rect = patches.Rectangle(
                (x1, y1), width, height,
                linewidth=2, edgecolor="red", facecolor="none"
            )
            ax.add_patch(rect)
            class_name = CLASS_MAPPING.get(int(cls_id), "desconhecido")
            plt.text(
                x1, y1 - 5, class_name,
                color="white", bbox=dict(facecolor="red", alpha=0.5)
            )
        plt.axis("off")
    plt.tight_layout()
    plt.show()

# Vamos inspecionar um batch do conjunto de validação
visualize_batch(val_ds, num_images=4)
```

## 3. Construção do Modelo YOLOV8Detector com KerasCV

Instanciei o detetor com *backbone* pré-treinado no COCO (`yolo_v8_s_backbone_coco`), configurando o otimizador e as perdas de classificação e localização.

### Seleção e Arquitetura do Modelo

Para atender aos exigentes requisitos industriais de inspeção óptica automatizada (AOI), optei pelo detector **YOLOv8 Small** (`yolo_v8_s_backbone_coco`) fornecido nativamente pelo ecossistema **KerasCV**. Os fatores determinantes para a minha escolha incluem:

*   **Eficiência de Passo Único (*Single-Stage*)**: Diferente de arquiteturas de dois passos (como Faster R-CNN), o YOLOv8 realiza a classificação e regressão de caixas delimitadoras de forma simultânea e unificada em uma única passagem de rede, reduzindo drasticamente o custo computacional.
*   **Estrutura do Backbone**: Utiliza uma arquitetura CSP (Cross Stage Partial) otimizada para extração profunda de características com caminhos de agregação de recursos multi-escala (*Path Aggregation Network* - PANet), permitindo capturar tanto pequenos detalhes geométricos de defeitos como *spur* e *mouse_bite* quanto estruturas maiores.
*   **Decoplagem de Cabeças (*Decoupled Head*)**: O YOLOv8 separa fisicamente os ramos que calculam as probabilidades de classe e as coordenadas de regressão geométrica. Isso mitiga o conflito clássico entre classificação e localização e acelera a convergência do modelo.
*   **Predição Sem Âncoras (*Anchor-Free*)**: O modelo prevê diretamente o centro do objeto em vez de depender de formatos pré-definidos de âncoras (anchor boxes). Isso confere flexibilidade extrema para se ajustar às dimensões exatas das anomalias de fabricação de PCBs.

### Por que utilizei os dados de pré-treino do COCO?

A utilização de pesos inicializados a partir do dataset **COCO (Common Objects in Context)** é uma estratégia fundamental de **Aprendizado por Transferência (*Transfer Learning*)** pelos seguintes motivos técnicos que considerei:

1.  **Extratores de Características Universais**: Redes neurais profundas aprendem de forma hierárquica. Camadas iniciais treinadas no COCO já aprenderam a detectar estruturas visuais primitivas cruciais (como bordas verticais/horizontais, curvas, texturas, transições de contraste e gradientes). Reutilizar esses filtros evita ter que treinar o modelo do zero (*from scratch*) para aprender o que é uma linha ou uma quina.
2.  **Aceleração drástica da Convergência**: Como o modelo começa o treinamento com filtros visuais altamente especializados e ricos, o processo de ajuste de pesos (*fine-tuning*) converge muito mais rápido. O modelo atinge perdas baixas e métricas de mAP estáveis em poucas épocas.
3.  **Mitigação de Overfitting (Sobreajuste)**: O conjunto de dados de defeitos de PCB, embora robusto, é pequeno se comparado aos milhões de parâmetros do YOLOv8. Treinar uma rede deste tamanho do zero em um conjunto restrito facilitaria a memorização do ruído das imagens de treino. Os pesos pré-treinados do COCO agem como um regularizador implícito extremamente poderoso.
4.  **Adaptação Eficiente ao Domínio de PCBs**: Embora uma placa de circuito (PCB) pareça visualmente diferente de objetos comuns do cotidiano (como pessoas, carros ou copos presentes no COCO), as anomalias industriais (furos, curtos, esporas) dependem fortemente da análise precisa de contornos, continuidade de linhas (trilhas) e contraste de formas geométricas simples — características que os filtros do COCO extraem com maestria.

```python
# Instanciação da arquitetura YOLOv8
model = keras_cv.models.YOLOV8Detector.from_preset(
    "yolo_v8_s_backbone_coco",
    num_classes=NUM_CLASSES,
    bounding_box_format="xyxy"
)

# Compilação com otimizador Adam e perdas dedicadas
model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-3),
    classification_loss="binary_crossentropy",
    box_loss="ciou"
)

model.summary()
```

## 4. Treino com Callbacks Obrigatórios

Conforme exigido nas diretrizes da disciplina, configurei:
- **`EarlyStopping`**: Interrompe o treino se a perda de validação estagnar, restaurando os pesos ideais.
- **`ModelCheckpoint`**: Guarda o artefacto apenas quando ocorre melhoria (`save_best_only=True`).
- **Extra - `PlotTrainingProgress`**: Plot do valor do erro durante o treinamento para monitoramento de métricas.

```python
from IPython.display import clear_output

class PlotTrainingProgress(keras.callbacks.Callback):
    """Callback dinâmico para plotar as perdas de treino e validação a cada época concluída."""
    def on_train_begin(self, logs=None):
        self.epochs = []
        self.losses = []
        self.val_losses = []

    def on_epoch_end(self, epoch, logs=None):
        self.epochs.append(epoch + 1)
        self.losses.append(logs.get('loss'))
        self.val_losses.append(logs.get('val_loss'))

        clear_output(wait=True)
        plt.figure(figsize=(10, 5))
        plt.plot(self.epochs, self.losses, label='Treino (Loss)', marker='o')
        if logs.get('val_loss') is not None:
            plt.plot(self.epochs, self.val_losses, label='Validação (Val Loss)', marker='s')
        plt.title(f'Evolução do Treinamento - Época {epoch + 1}')
        plt.xlabel('Época')
        plt.ylabel('Loss')
        plt.legend()
        plt.grid(True)
        plt.show()

# Caminho de salvamento do melhor modelo no Google Drive
checkpoint_drive_filepath = os.path.join(DRIVE_PROJECT_PATH, "checkpoints/best_pcb_yolo.keras")

# Configuração dos Callbacks atualizada para salvar no Drive
callbacks = [
    EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True,
        verbose=1
    ),
    ModelCheckpoint(
        filepath=checkpoint_drive_filepath,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),
    PlotTrainingProgress()
]

print(f"Checkpoints configurados para salvar em: {checkpoint_drive_filepath}")
```

### 4.1 Execução Estratégica do Treinamento

Esta célula realiza o treinamento efetivo da rede YOLOv8 adotando defesas cruciais que planejei para a estabilidade do ambiente no Google Colab:
- **Prevenção de OOM (Out Of Memory)**: Limpei explicitamente a sessão do Keras e redefini o `BATCH_SIZE` para `4` de modo a limitar a quantidade de imagens processadas simultaneamente na GPU.
- **Desativação do JIT (Just-In-Time) no Keras 3**: Configurei `jit_compile=False` na compilação do modelo para impedir que otimizações estáticas de grafos travem o processo ou corrompam a gravação contínua do arquivo de checkpoints.
- **Ajuste de Passos e Épocas**: Calculei o total de amostras de treino para ajustar a variável `STEPS_PER_EPOCH` de maneira proporcional ao novo tamanho de lote, garantindo que cada época utilize o conjunto total de dados.

```python
import os
import glob
import tensorflow as tf
import keras
import keras_cv
from keras import optimizers
from keras.callbacks import EarlyStopping, ModelCheckpoint

# Forçar a limpeza de memória da GPU antes de começar
tf.keras.backend.clear_session()

# 1. Ajuste estratégico do Batch Size para evitar OOM (Out Of Memory)
BATCH_SIZE = 4

# Recriando os pipelines com o novo BATCH_SIZE reduzido
train_ds = prepare_pipeline(load_dataset_split(train_img_dir, train_lbl_dir), batch_size=BATCH_SIZE, is_training=True)
val_ds = prepare_pipeline(load_dataset_split(val_img_dir, val_lbl_dir), batch_size=BATCH_SIZE, is_training=False)
test_ds = prepare_pipeline(load_dataset_split(test_img_dir, test_lbl_dir), batch_size=BATCH_SIZE, is_training=False)

os.makedirs("models", exist_ok=True)

# Re-compilação limpa sem JIT
model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-3),
    classification_loss="binary_crossentropy",
    box_loss="ciou",
    jit_compile=False
)

# Recalcula a quantidade de passos por época baseado no novo BATCH_SIZE
NUM_TRAIN_SAMPLES = len(glob.glob(os.path.join(train_img_dir, "*.[jJ][pP][gG]")))
STEPS_PER_EPOCH = NUM_TRAIN_SAMPLES // BATCH_SIZE

callbacks = [
    EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True,
        verbose=1
    ),
    ModelCheckpoint(
        filepath="models/best_pcb_yolo.keras",
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),
    PlotTrainingProgress() # Callback dinâmico
]

print(f"Treinamento reiniciado estrategicamente com Batch Size = {BATCH_SIZE} para evitar travamentos.")

# Execução de Treino Otimizada
history = model.fit(
     train_ds,
     validation_data=val_ds,
     epochs=EPOCHS,
     steps_per_epoch=STEPS_PER_EPOCH,
     callbacks=callbacks
 )
```

## 7. Avaliação Final e Cálculo do mAP no Conjunto de Teste

Executei o conjunto de teste apenas uma única vez no final da modelagem com o `BoxMeanAveragePrecision`.

```python
import sys
import subprocess

# Garante a instalação do keras_cv no ambiente caso tenha sido perdido pelo reset do kernel
try:
    import keras_cv
except ModuleNotFoundError:
    print("Instalando KerasCV silenciosamente...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "keras_cv"])
    import keras_cv

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# 1. Configurações e Limiares de Decisão
IOU_THRESHOLD = 0.5
CONF_THRESHOLD = 0.25
NUM_CLASSES = 6

CLASS_MAPPING = {
    0: "missing_hole",
    1: "mouse_bite",
    2: "open_circuit",
    3: "short",
    4: "spur",
    5: "spurious_copper"
}
class_names = [CLASS_MAPPING[i] for i in range(NUM_CLASSES)]

# Estruturas para acumular TP, FP e FN
# Classe 'NUM_CLASSES' (índice 6) representará o "Fundo" (Background)
confusion_data = np.zeros((NUM_CLASSES + 1, NUM_CLASSES + 1), dtype=np.int64)

def compute_iou(box1, box2):
    """Calcula o IoU entre duas caixas no formato [x1, y1, x2, y2]."""
    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])
    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    intersection = max(0.0, x2 - x1) * max(0.0, y2 - y1)
    area1 = (box1[2] - box1[0]) * (box1[3] - box1[1])
    area2 = (box2[2] - box2[0]) * (box2[3] - box2[1])
    union = area1 + area2 - intersection

    return intersection / union if union > 0 else 0.0

# Carrega o modelo se ele já não estiver em memória
if 'loaded_model' not in globals():
    model_save_path = os.path.join(DRIVE_PROJECT_PATH, "models/pcb_yolo_final.keras")
    print(f"Carregando o modelo para avaliação: {model_save_path} ...")
    loaded_model = tf.keras.models.load_model(model_save_path, safe_mode=False)

print("Processando predições no conjunto de teste para calcular a matriz de confusão...")

for batch in test_ds:
    images = batch["images"]
    y_true = batch["bounding_boxes"]
    y_pred = loaded_model.predict(images, verbose=0)

    # Conversão de batches para numpy
    true_boxes_batch = y_true["boxes"].to_list() if isinstance(y_true["boxes"], tf.RaggedTensor) else y_true["boxes"].numpy()
    true_classes_batch = y_true["classes"].to_list() if isinstance(y_true["classes"], tf.RaggedTensor) else y_true["classes"].numpy()

    pred_boxes_batch = y_pred["boxes"]
    pred_classes_batch = y_pred["classes"]
    pred_conf_batch = y_pred["confidence"]

    for idx in range(len(images)):
        # Filtra elementos reais legítimos (ignora padding -1)
        t_boxes = [b for b, c in zip(true_boxes_batch[idx], true_classes_batch[idx]) if c != -1]
        t_classes = [int(c) for c in true_classes_batch[idx] if c != -1]

        # Filtra predições baseadas no limiar de confiança
        p_boxes = []
        p_classes = []
        p_confs = []
        for b, c, conf in zip(pred_boxes_batch[idx], pred_classes_batch[idx], pred_conf_batch[idx]):
            if c != -1 and conf >= CONF_THRESHOLD:
                p_boxes.append(b)
                p_classes.append(int(c))
                p_confs.append(conf)

        detected_true_indices = set()

        # Para cada predição, busca a melhor caixa real correspondente
        for p_idx, (p_box, p_cls) in enumerate(zip(p_boxes, p_classes)):
            best_iou = 0.0
            best_t_idx = -1

            for t_idx, (t_box, t_cls) in enumerate(zip(t_boxes, t_classes)):
                if t_idx in detected_true_indices:
                    continue
                iou = compute_iou(p_box, t_box)
                if iou > best_iou:
                    best_iou = iou
                    best_t_idx = t_idx

            if best_iou >= IOU_THRESHOLD and best_t_idx != -1:
                # Encontrou correspondência geométrica
                matched_class = t_classes[best_t_idx]
                confusion_data[matched_class, p_cls] += 1
                detected_true_indices.add(best_t_idx)
            else:
                # Falso Positivo: Predição que não corresponde a nenhum defeito real (Background)
                confusion_data[NUM_CLASSES, p_cls] += 1

        # Falsos Negativos: Defeitos reais que não foram detectados pelo modelo
        for t_idx, t_cls in enumerate(t_classes):
            if t_idx not in detected_true_indices:
                confusion_data[t_cls, NUM_CLASSES] += 1

# 2. Plots de Avaliação
plt.figure(figsize=(12, 10))

# Normalização das linhas para percentual de acerto/erro por classe real
with np.errstate(all='ignore'):
    confusion_data_normalized = confusion_data.astype('float') / confusion_data.sum(axis=1)[:, np.newaxis]
    confusion_data_normalized = np.nan_to_num(confusion_data_normalized)

# Gerar labels para as classes incluindo Background
labels_extended = class_names + ["background"]

# Plot 1: Matriz de Confusão
sns.heatmap(
    confusion_data_normalized,
    annot=confusion_data,
    fmt="d",
    cmap="Blues",
    xticklabels=labels_extended,
    yticklabels=labels_extended
)
plt.title("Matriz de Confusão - Deteção de Defeitos de PCB\n(Valores absolutos com normalização de linha)", fontsize=14)
plt.ylabel("Classe Real (Ground Truth)", fontsize=12)
plt.xlabel("Classe Predita pelo YOLOv8", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# Plot 2: Métricas Rápidas por Classe
print("\n--- Relatório Métrico Operacional (IoU >= 0.5) ---")
for i in range(NUM_CLASSES):
    tp = confusion_data[i, i]
    fp = confusion_data[NUM_CLASSES, i] + sum(confusion_data[j, i] for j in range(NUM_CLASSES) if j != i)
    fn = confusion_data[i, NUM_CLASSES] + sum(confusion_data[i, j] for j in range(NUM_CLASSES) if j != i)

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0

    print(f"{class_names[i]:<20} | Precisão: {precision:.4f} | Sensibilidade (Recall): {recall:.4f} | F1-Score: {f1:.4f}")
```

Por fim, este projeto consolidou a minha aplicação prática de múltiplos conceitos essenciais abordados nos **Blocos 1, 2 e 3** do curso:
*   **Bloco 1 (Fundamentos e Pré-processamento)**: Minha manipulação estruturada de tensores multidimensionais, carregamento dinâmico via geradores no `tf.data.Dataset`, normalização de imagens e conversão de caixas delimitadoras absolutas para adequação ao formato exigido por bibliotecas modernas de Visão Computacional.
*   **Bloco 2 (Modelagem Deep Learning)**: Meu uso de aprendizado por transferência (*Transfer Learning*) utilizando um backbone pré-treinado no conjunto COCO, compilação customizada com funções de perda multi-tarefa (perda geométrica `ciou` combinada com classificação por entropia cruzada binária) e implementação de callbacks críticos de monitoramento como o `EarlyStopping` e o `ModelCheckpoint`.
*   **Bloco 3 (Otimização e Implantação)**: Minha exportação otimizada do grafo computacional, quantização dinâmica e conversão de formato de precisão para geração de arquivos binários leves no formato `.tflite`, além da documentação formal de limitações, escopo e requisitos de uso do modelo sob os princípios éticos e técnicos de um **Model Card** profissional.
